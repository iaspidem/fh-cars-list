```python
import pandas as pd
```

```python
filename = "fh6cars-15-07-26"
```

```python
tables = pd.read_html(filename + ".htm")
df = tables[1]
df.head()
```

```python
df.info()
```

```python
columns_to_drop = [1, 2, 3, 4]
df = df.drop(df.columns[columns_to_drop], axis=1)
df.head()
```

```python
df = df.rename(columns={"Sp": "Speed", "Ha": "Handling", "Ac": "Accel", "La": "Launch", "Br": "Braking", "Of": "OffRd"})
df.head()
```

```python
df.info()
```

```python
columns_to_drop = [2, 3, 4, 5, 6, 7]
df_lite = df.drop(df.columns[columns_to_drop], axis=1)
df_lite.head()
```

```python
df_lite[["Value", "Rarity"]] = df["Value"].str.split("CR", n=1, expand=True)
df_lite = df_lite.iloc[:, [0, 1, 3, 2]]

df_lite["Value"] = df_lite["Value"].str.replace(",", "")
df_lite["Value"] = df_lite["Value"].str.replace("?", "0")
df_lite["Value"] = df_lite["Value"].astype(int)
df_lite.head()
```

```python
df_lite.info()
```

```python
df_sort = df_lite[df_lite["Vehicle"].str.contains("Autoshow, Wheelspin")]
df_sort.head()
```

```python
df_sort.info()
```

```python
df_sort = df_sort.sort_values(["Vehicle", "Value"], ascending=[True, True])
df_sort.head()
```

```python
df_sort.to_csv(filename + "-sorted" + ".csv", index=False)
```

```python
df_lite.to_csv(filename + ".csv", index=False)
```
