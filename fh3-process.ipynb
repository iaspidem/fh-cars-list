# Import raw data for preprocessing

```python
filename = "raw-website/FH3-cars-list"
tables = pd.read_html(filename + ".htm")
df = tables[1]
df.head()
```

## Remove redundant columns

```python
columns_to_drop = [1]
df = df.drop(df.columns[columns_to_drop], axis=1)
df.head()
```

## Expand abbreviated columns

```python
df = df.rename(columns={"Sp": "Speed", "Ha": "Handling", "Ac": "Accel", "La": "Launch", "Br": "Braking", "Of": "OffRd"})
df.head()
```

```python
df.info()
```

## Remove stat columns from df_lite

```python
columns_to_drop = [2, 3, 4, 5, 6]
df_lite = df.drop(df.columns[columns_to_drop], axis=1)
df_lite.head()
```

## Split original Value column into Value and Car Type

```python
df_lite[["Value", "Car Type"]] = df["Value"].str.split("CR", n=1, expand=True)
df_lite = df_lite.iloc[:, [0, 1, 3, 2]]

df_lite["Value"] = df_lite["Value"].str.replace(",", "")
df_lite["Value"] = df_lite["Value"].str.replace("?", "0")
df_lite["Value"] = df_lite["Value"].astype(int)
df_lite.head()
```

## Split original Vehicle column into Brand and Vehicle

```python
df_lite[["Brand", "Vehicle"]] = df_lite["Vehicle"].str.split(" ", n=1, expand=True)
df_lite = df_lite.iloc[:, [4, 0, 1, 2, 3]]
df_lite.head()
```

## Split Vehicle column into Vehicle and Collection

```python
df_lite[["Vehicle", "Collection"]] = df_lite["Vehicle"].str.split(r'(?=Autoshow)|(?=Barn)|(?=Wheelspin)|(?=Forzathon)|(?=Gifted)', n=1,
                                                                  expand=True)
df_lite.head()
```

```python
df_lite["Collection"].value_counts()
```

## Split Vehicle Column into Year and Vehicle

```python
df_lite[["Vehicle", "Year"]] = df_lite["Vehicle"].str.rsplit(" ", n=1, expand=True)
df_lite = df_lite.iloc[:, [0, 6, 1, 2, 3, 5, 4]]
df_lite.head()
```
