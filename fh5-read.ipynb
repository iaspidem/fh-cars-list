```python
import pandas as pd
```

```python
df = pd.read_csv("FH5/FH5-0-owned.csv", encoding="cp1252")
df.head()
```

```python
df.dtypes
```

```python
df["Year"] = df["Year"].astype(str)
df.dtypes
```

```python
df["Owned"] = df["Owned"].fillna("1")
df.head()
```

```python
df["Owned"].value_counts()
```

```python
df = df.sort_values(["Brand", "Year", "Vehicle"], ascending=[True, True, True], key=lambda col: col.str.lower())
df = df.reset_index(drop=True)
df.head()
```

```python
df_notowned = df.iloc[df["Owned"].str.contains("0")]
df_notowned.head()
```

```python
df_notowned_htf = df_notowned[df_notowned["Collection"].str.contains("Hard to Find")]
df_notowned_htf.head()
```

```python
df_save = df_notowned_htf.copy()
```

```python
df_save["Vehicle"] = df_notowned_htf["Brand"] + " " + df_notowned_htf["Vehicle"] + "(" + df_notowned_htf["Year"] + ")"
df_save.head()
```

```python
columns_to_drop = [0, 2]
df_save = df_save.drop(df_save.columns[columns_to_drop], axis=1)
df_save.head()
```

```python
df_nonautoshow = df_notowned[~df_notowned["Collection"].str.contains("Autoshow")]
df_nonautoshow["Vehicle"] = df_nonautoshow["Year"] + " " + df_nonautoshow["Brand"] + " " + df_nonautoshow["Vehicle"]
df_nonautoshow = df_nonautoshow.drop(df_nonautoshow.columns[columns_to_drop], axis=1)
df_nonautoshow.head()
```
