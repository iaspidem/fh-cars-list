```python
import pandas as pd
```

# Import raw data for preprocessing

```python
filename = "FH4-cars-list"
tables = pd.read_html("raw-website/" + filename + ".htm")
df = tables[1]
df.head(10)
```

## Remove redundant columns

```python
columns_to_drop = [0, 2, 3, 4]
df = df.drop(df.columns[columns_to_drop], axis=1)
df.head()
```

## Expand abbreviated columns

```python
df = df.rename(columns={"Sp": "Speed", "Ha": "Handling", "Ac": "Accel", "La": "Launch", "Br": "Braking", "Of": "OffRd"})
df.head()
```

```python
df.info()
```

## Remove stat columns from df_lite

```python
columns_to_drop = [2, 3, 4, 5, 6]
df_lite = df.drop(df.columns[columns_to_drop], axis=1)
df_lite.head()
```

## Split original Value column into Value and Rarity

```python
df_lite.iloc[df_lite["Value"].str.contains("?", regex=False)].head()
```

```python
df_lite[["Value", "Rarity"]] = df["Value"].str.split("CR", n=1, expand=True)
df_lite = df_lite.iloc[:, [0, 1, 3, 2]]
df_lite.head()
```

### Strip leading and trailing whitespace

```python
df_lite["Value"] = df_lite["Value"].str.strip()
df_lite["Rarity"] = df_lite["Rarity"].str.strip()
df_lite["Value"] = df_lite["Value"].str.replace(",", "")
df_lite["Value"] = df_lite["Value"].astype(int)
df_lite.head()
```

## Split original Vehicle column into Brand and Vehicle

```python
df_lite.iloc[df_lite["Vehicle"].str.contains("%")].head()
```

### Isolate rows for 2-word vehicle brand names

```python
df1 = df_lite.iloc[df_lite["Vehicle"].str.contains("Alfa Romeo|Alumi Craft|Aston Martin|Auto Union|DS Automobiles|Formula Drift|Funco Motorsports|Hot Wheels|Land Rover|Local Motors|RJ Anderson|Spania GTA|Top Gear|W Motors")]
df1 = df1.loc[~df1["Vehicle"].str.contains("Mercedes-Benz|James Bond Edition")]
df1.head()
```

```python
df1["Vehicle"] = df1["Vehicle"].str.replace(" ", "%", n=1)
df1.head()
```

### Isolate rows for 3-word vehicle brand names

```python
df2 = df_lite.iloc[df_lite["Vehicle"].str.contains("AMG Transport|James Bond Edition|LEGO Speed Champions")]
df2.head()
```

```python
df2["Vehicle"] = df2["Vehicle"].str.replace(" ", "%", n=2)
df2.head()
```

```python
df_lite.update(df1)
df_lite.head()
```

```python
df_lite.update(df2)
df_lite.head()
```

### Perform split

```python
df_lite[["Brand", "Vehicle"]] = df_lite["Vehicle"].str.split(" ", n=1, expand=True)
df_lite = df_lite.iloc[:, [4, 0, 1, 2, 3]]
df_lite.head()
```

#### Error checking

```python
df_verify = df_lite.iloc[df_lite["Vehicle"].str.contains("%")]
df_verify.head()
```

#### Replace percent sign in Brand column with whitespace

```python
df_lite["Brand"] = df_lite["Brand"].str.replace("%", " ")
df_lite.head()
```

## Split Vehicle column into Vehicle and Collection

```python
df_lite[["Vehicle", "Year", "Collection"]] = df_lite["Vehicle"].str.extract(r'^(.*)(\d{4})(.*)$')
df_lite = df_lite.iloc[:, [0, 1, 5, 2, 6, 3, 4]]
df_lite.loc[644, "Vehicle"] = "Regalia"
df_lite.loc[645, "Vehicle"] = "Regalia Type-D"
df_lite.loc[[644, 645], "Year"] = "723"
df_lite.loc[644, "Collection"] = "A \"Hard-to-Find\" vehicle"
df_lite.loc[645, "Collection"] = "Skill Tree:  Quartz Regalia"
df_lite.head()
```

```python
df_lite["Collection"].value_counts()
```

### Strip leading and trailing whitespace

```python
df_lite["Collection"] = df_lite["Collection"].str.strip()
```

## Sort by Brand then Year and Vehicle (ascending)

```python
df_lite = df_lite.sort_values(["Brand", "Year", "Vehicle"], ascending=[True, True, True], key=lambda col: col.str.lower())
df_lite = df_lite.reset_index(drop=True)
df_lite.head(10)
```

## Clean PI column

```python
df_lite[["PI1", "PI2"]] = df_lite["PI"].str.split(r'(?<=D)|(?<=C)|(?<=B)|(?<=A)|(?<=S1)|(?<=S2)', n=1,
                                                                  expand=True)
df_lite.head()
```

```python
df_lite["PI1"].value_counts()
# "S1" is the most common class
# followed by A
# then D
```

```python
df_lite["PI2"].value_counts()
# D 100 is the most common PI value
```

```python
df_lite["PI"] = df_lite["PI1"] + " " + df_lite["PI2"]
df_lite = df_lite.drop(["PI1", "PI2"], axis="columns")
df_lite.head()
```

```python
df_lite["Owned"] = pd.Series(dtype="string")
df_lite.head()
```

## Filter to Wheelspin-exclusive cars

```python
df_lite.dtypes
```

```python
df_wheelspin = df_lite.loc[df_lite["Collection"].str.contains("Wheelspin")]
df_wheelspin.head()
```

```python
df_wheelspin["Rarity"].value_counts()
```

### Associate each rarity with numerical value

```python
rarity_dict = {"COMMON": 1, "RARE": 2, "EPIC": 3, "LEGENDARY": 4, "FORZA EDITION": 5}
df_wheelspin["Rarity Value"] = df_wheelspin["Rarity"].map(rarity_dict)
df_wheelspin.head()
```

### Sort Wheelspin-exclusives according to Rarity

```python
df_wheelspin = df_wheelspin.sort_values(["Rarity Value", "Brand"], ascending=[True, True])
df_wheelspin.head()
```

```python
df_wheelspin = df_wheelspin.drop("Rarity Value", axis=1)
df_wheelspin.head()
```
